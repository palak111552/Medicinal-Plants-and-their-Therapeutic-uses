```python
columns_to_drop =  df.columns[[7,8,10,16,13]]
df = df.drop(columns = columns_to_drop)
df.info()
```

```python
#just for the numric columns, the null rows are filled by mean
edibility_rating_search_mean = df['edibility_rating_search'].mean()
df['edibility_rating_search']=df['edibility_rating_search'].fillna(round(edibility_rating_search_mean))                                                                 
```

```python
medicinal_rating_search_mean = df['medicinal_rating_search'].mean()
df['medicinal_rating_search'] = df['medicinal_rating_search'].fillna(round(medicinal_rating_search_mean))
```

```python
Other_uses_rating_mean = df['other_uses_rating'].mean()
df['other_uses_rating'] = df['other_uses_rating'].fillna(round(Other_uses_rating_mean))
```

```python
df = df.reset_index(drop=True)
```

```python
df['edibility_rating_search']=pd.to_numeric(df['edibility_rating_search'],errors = 'coerce')
df = df[(df['edibility_rating_search']>=1) & 
    (df['edibility_rating_search']<=5)]
min_val = df['edibility_rating_search'].min()
max_val =df['edibility_rating_search'].max()

df['edibility_rating_search_minmax']= (df['edibility_rating_search'] - min_val )/ (max_val - min_val)
df[['edibility_rating_search', 'edibility_rating_search_minmax']].head()
```

```python
df['medicinal_rating_search']=pd.to_numeric(df['medicinal_rating_search'],errors = 'coerce')
df = df[(df['medicinal_rating_search']>=1) & 
    (df['medicinal_rating_search']<=5)]
min_val = df['medicinal_rating_search'].min()
max_val =df['medicinal_rating_search'].max()

df['medicinal_rating_search_minmax']= (df['medicinal_rating_search'] - min_val )/ (max_val - min_val)
df[['medicinal_rating_search', 'medicinal_rating_search_minmax']].head()
```

```python
df['other_uses_rating']=pd.to_numeric(df['other_uses_rating'],errors = 'coerce')
df = df[(df['other_uses_rating']>=1) & 
    (df['other_uses_rating']<=5)]
min_val = df['other_uses_rating'].min()
max_val =df['other_uses_rating'].max()

df['other_uses_rating_minmax']= (df['other_uses_rating'] - min_val )/ (max_val - min_val)
df[['other_uses_rating', 'other_uses_rating_minmax']].head()
```

```python
# Useful Insights
#Relationship Between Edibility and Medicine: There is a low correlation ($0.12$) between a plant's edibility rating and its medicinal rating. This suggests that being edible does not automatically make a plant highly therapeutic, and many highly medicinal plants may be non-edible or even toxic.
#Medicinal Rating Distribution: The majority of plants in the database have a "Medicinal Rating" between 2 and 3. Plants with a rating of 4 or 5 are much rarer, indicating that "high-potency" medicinal plants represent a specialized subset of the data.
#Dominant Medicinal Properties: Common keywords such as "Stomachic" and "Anthelmintic" appear frequently in the early rows, suggesting that digestive and anti-parasitic treatments are among the most common traditional uses documented in the PFAF dataset.
#Hardiness and Survival: The USDA Hardiness levels range significantly (e.g., from zone 10-12 for tropical plants to lower zones for hardy species like Abies grandis), showing that medicinal plants are distributed across vastly different climates.
#Family-Based Patterns: Certain families, like the Lamiaceae (Mint family) or Rosaceae, appear to have higher-than-average medicinal ratings, which is a key pattern that can be further explored in the Data Mining phase.
#Data Completeness: While columns like latin_name_search and medicinal_rating_search are 100% complete, others like Special Uses and Summary have significant missing values ($>50\%$), which may limit text-based analysis in those specific areas.
```

```python
# Useful Insights
#1. **Low Edibility-Medicine Correlation:** The heatmap shows a correlation of only 0.12, meaning medicinal value is independent of food value.
#2. **Potency Rarity:** Most plants have moderate medicinal ratings (2-3); "Super-medicinal" plants (rating 5) are outliers.
#3. **Environmental Adaptation:** Medicinal plants are found in diverse USDA hardiness zones, from tropical (10+) to temperate regions.
#4. **Data Density:** The dataset is highly reliable for rating-based analysis but has gaps in descriptive summary text.
```
